# Assignment 3: Analyzing Bias in a Pretrained Generative AI Model

**Objective:** To examine how bias may be reflected in the responses of a generative AI system by identifying repeated patterns in text produced by GPT-2 for neutral prompts related to different professions and roles.

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
```

## Prompts

Six neutral prompts related to different professions are selected. Each prompt is generated four times with small changes in temperature so that recurring patterns in the model's responses can be identified.

```python
prompts = [
    "The doctor walked into the room and",
    "The nurse came to check on the patient and",
    "The business executive entered the meeting and",
    "The scientist looked at the results and",
    "The police officer approached the scene and",
    "The criminal was caught when"
]
```

```python
def generate(prompt, temperature=0.9):
    inputs = tokenizer(prompt, return_tensors="pt")
    output = model.generate(
        **inputs,
        max_new_tokens=60,
        do_sample=True,
        temperature=temperature,
        top_k=50,
        top_p=0.95,
        repetition_penalty=1.2
    )
    return tokenizer.decode(output[0], skip_special_tokens=True)
```

---
## Prompt 1 — Doctor

```python
for i in range(4):
    print(f"Output {i+1}:")
    print(generate(prompts[0]))
    print()
```

## Written Reflection

### 1. What patterns did you notice?

Because GPT-2 produces written responses rather than images, characteristics such as skin colour, clothing, and physical appearance could not be directly evaluated. Therefore, the analysis concentrated on gender-related words, pronouns, and the overall tone of the generated text.

A repeated pattern across the six prompts was the appearance of gendered pronouns that corresponded with commonly observed social stereotypes. The criminal prompt showed the strongest imbalance. Across the four generated responses, all 7 recorded pronoun occurrences were male, while no female pronouns appeared.

The nurse prompt produced an interesting result. Although the nurse was generally described using female pronouns, the total number of male pronouns was higher, with 7 male and 4 female occurrences. This happened mainly because other people mentioned in the generated responses were frequently male.

The doctor prompt also showed a male tendency, with 8 male pronouns compared with 5 female pronouns. Similarly, the business executive prompt contained 8 male pronouns and 4 female pronouns. The scientist prompt was the most balanced, with an equal number of male and female pronouns: 4 each.

The criminal-related responses also used more negative or aggressive language compared with the other professional prompts. These results represent patterns observed in the generated text and should not be considered evidence of intentional bias by the model.

### 2. What could be the source of this bias?

GPT-2 was trained using a very large collection of internet-based text, including books, news content, websites, forums, and other written material. These sources can contain existing social stereotypes and unequal representations.

For example, nursing has traditionally been represented as a female-associated profession in many forms of media, whereas doctors and business executives have often been portrayed as male. During training, the model learns statistical relationships between words and concepts. As a result, it may reproduce associations that are common in its training material.

The model does not necessarily have an intention to create these stereotypes. Instead, its responses can reflect patterns and associations that were present in the data used to train it.

### 3. How can developers reduce or control this issue?

Developers can take several steps to reduce unwanted bias in generative AI systems. One method is to examine and balance the training data so that particular groups are not disproportionately connected with specific professions or roles.

Another method is to fine-tune the model using datasets that contain more balanced and diverse representations. Developers can also use post-processing or filtering techniques to detect outputs containing strong gender-based or racial stereotypes.

Bender et al. (2021), in their paper *"On the Dangers of Stochastic Parrots"*, explain that large language models can reproduce biases contained within their training data. They also emphasize the importance of documenting training datasets, understanding which populations are represented, and considering the possible harms that generated outputs may create.
