## Objective

The objective of this assignment is to understand how different text-generation parameters influence the output produced by GPT-2.

```python
import torch
from transformers import pipeline, set_seed
```

```python
# Load the pretrained GPT-2 model
generator = pipeline(
    "text-generation",
    model="openai-community/gpt2"
)

# Set seed for reproducible results
set_seed(42)

print("GPT-2 model loaded successfully!")
```

```python
prompt = "The future of artificial intelligence will"

print("Original Prompt:")
print(prompt)
```

```python
def generate_text(prompt, temperature=1.0, top_k=50, top_p=1.0, repetition_penalty=1.0):

    output = generator(
        prompt,
        max_new_tokens=60,
        do_sample=True,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p,
        repetition_penalty=repetition_penalty,
        num_return_sequences=1
    )

    return output[0]["generated_text"]
```

```python
set_seed(42)

output_1 = generate_text(
    prompt,
    temperature=0.3,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0
)

print("EXPERIMENT 1")
print("Parameter changed: Temperature")
print("Value: 0.3")
print()
print(output_1)
```

```python
set_seed(42)

output_2 = generate_text(
    prompt,
    temperature=1.2,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0
)

print("EXPERIMENT 2")
print("Parameter changed: Temperature")
print("Value: 1.2")
print()
print(output_2)
```

```python
set_seed(42)

output_3 = generate_text(
    prompt,
    temperature=1.0,
    top_k=10,
    top_p=1.0,
    repetition_penalty=1.0
)

print("EXPERIMENT 3")
print("Parameter changed: Top-k")
print("Value: 10")
print()
print(output_3)
```

```python
set_seed(42)

output_4 = generate_text(
    prompt,
    temperature=1.0,
    top_k=50,
    top_p=0.5,
    repetition_penalty=1.0
)

print("EXPERIMENT 4")
print("Parameter changed: Top-p")
print("Value: 0.5")
print()
print(output_4)
```

```python
set_seed(42)

output_5 = generate_text(
    prompt,
    temperature=1.0,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.5
)

print("EXPERIMENT 5")
print("Parameter changed: Repetition Penalty")
print("Value: 1.5")
print()
print(output_5)
```

## Comparison of Five Experiments

| Experiment | Parameter Changed | Value | Observation |
|---|---|---:|---|
| 1 | Temperature | 0.3 | Output was more predictable and focused. |
| 2 | Temperature | 1.2 | Output was more creative and random. |
| 3 | Top-k | 10 | Output was restricted to a smaller set of likely tokens. |
| 4 | Top-p | 0.5 | Output was more controlled by limiting the probability range. |
| 5 | Repetition Penalty | 1.5 | Repeated words and phrases were discouraged. |


```python
print("=" * 70)
print("EXPERIMENT 1 - TEMPERATURE = 0.3")
print("=" * 70)
print(output_1)

print("\n")

print("=" * 70)
print("EXPERIMENT 2 - TEMPERATURE = 1.2")
print("=" * 70)
print(output_2)

print("\n")

print("=" * 70)
print("EXPERIMENT 3 - TOP-K = 10")
print("=" * 70)
print(output_3)

print("\n")

print("=" * 70)
print("EXPERIMENT 4 - TOP-P = 0.5")
print("=" * 70)
print(output_4)

print("\n")

print("=" * 70)
print("EXPERIMENT 5 - REPETITION PENALTY = 1.5")
print("=" * 70)
print(output_5)
```

## Analysis of the Five Experiments

The experiment with a low temperature produced more predictable and focused text. When the temperature was increased, the generated text became more varied and creative, although it could also become less coherent.

The lower top-k value restricted the model to a smaller group of likely words, making the generation more controlled. Similarly, a lower top-p value limited the probability range of possible tokens.

The repetition penalty helped reduce repeated words or phrases. Overall, temperature had a noticeable effect on the randomness of the generated text, while top-k and top-p controlled which possible tokens were considered during generation.

## Observations from Custom Generations

The same prompt was used for all three generations, but the temperature was changed.

At temperature 0.5, the generated text was relatively predictable and focused. At temperature 0.9, the output showed more variety while still remaining reasonably coherent. At temperature 1.3, the output was more creative and unpredictable, although some parts could become less meaningful.

This experiment shows that changing the generation parameters can significantly affect the style and randomness of GPT-2's output.

## Additional Activity: High Temperature Output

At a high temperature, the model produced a more unusual and less predictable output. This happened because a higher temperature increases the randomness of token selection. As a result, GPT-2 was more likely to choose lower-probability words, which can sometimes reduce the coherence of the generated text.

## Conclusion

This experiment helped me understand how GPT-2 generates text and how generation parameters affect its output. Lower temperature generally produced more predictable text, while higher temperature produced more creative and random text. Top-k and top-p controlled the range of possible tokens considered during generation. Repetition penalty helped reduce repeated words and phrases. Therefore, generation parameters can be adjusted depending on whether we want predictable, controlled, or creative text.
